```python
import pandas as pd

data = response.json()

data.keys()
```

```python
name = data['name']
birth_date = data['birthDate']['utcTime'][:10]
team = data['primaryTeam']['teamName']

rows = pd.DataFrame({'title': ['Name', 'Birth Date', 'Team'], 'value.fallback': [name, birth_date, team]})

details_data = data['playerInformation']

details_norm_df = pd.json_normalize(details_data)
details_df = details_norm_df[['title', 'value.fallback']]

details_df = pd.concat([rows, details_df])

details_df
```

```python
pos_data = data['positionDescription']['positions']

pos_norm_df = pd.json_normalize(pos_data)

pos_df = pos_norm_df.iloc[:,:4]

pos_df
```

```python
heatmap_data = data['firstSeasonStats']['heatmap']['coordinates']

heatmap_df = pd.DataFrame(heatmap_data)
```

```python
shotmap_data = data['firstSeasonStats']['shotmap']

shotmap_df = pd.DataFrame(shotmap_data)

shotmap_df.loc[(shotmap_df['eventType'] == 'AttemptSaved') & (shotmap_df['isBlocked'] == True), 'eventType'] = 'Blocked'

shotmap_df
```

```python
# shotmap_df.to_excel('dt_shotmap.xlsx', index=False)
```

```python
stats_data = data['firstSeasonStats']['statsSection']['items']

stats_df = pd.DataFrame()

for i in range(len(stats_data)):
  stats_df = pd.concat([stats_df, pd.DataFrame(stats_data[i]['items'])], ignore_index=True)

stats_df = stats_df.sort_values(['percentileRankPer90', 'percentileRank'], ascending=False)

stats_df = stats_df.loc[stats_df['percentileRankPer90'] > 65]

stats_df
```

```python
bundesliga_2324 = sb.matches(competition_id=9, season_id=281)
```

```python
bundesliga_2324_match_ids = list(bundesliga_2324['match_id'])
```

```python
wsl_2324 = sb.matches(37,281)
```

```python
wsl_2324_match_ids = list(wsl_2324['match_id'])
```

```python
events_list = []
for id in wsl_2324_match_ids:
  temp_events = sb.events(id)
  events_list.append(temp_events)
  print(f'Progress: {wsl_2324_match_ids.index(id) + 1}/{len(wsl_2324_match_ids)}')

wsl_2324_events = pd.concat(events_list)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
injury_url = 'https://www.transfermarkt.co.uk/jonny-yull/verletzungen/spieler/857060'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36'}

response = requests.get(injury_url, headers=headers)
response.status_code
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
html_table = soup.select_one('div[class="responsive-table"]')
```

```python
html_table
```

```python
injury_df = pd.read_html(str(html_table))[0]
```

```python
injury_df.to_excel('jy_injury_df.xlsx', index=False)
```

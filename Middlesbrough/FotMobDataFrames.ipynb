```python
data = response.json()
```

```python
player_data = data['players']
top_scorers = player_data[0]['topThree']
names = []
ranks = []
countries = []
values = []
stat_names = []

for i in range(len(player_data)):
    top_three = player_data[i]['topThree']
    for j in range(len(top_three)):
        names.append(top_three[j]['name'])
        ranks.append(top_three[j]['rank'])
        countries.append(top_three[j]['ccode'])
        values.append(top_three[j]['value'])
        stat_names.append(player_data[i]['header'])

player_df = pd.DataFrame({'Name': names,
             'League Rank': ranks,
             'Country': countries,
             'Value': values,
             'Stat Name': stat_names})
```

```python
team_data = data['teams']

names = []
ranks = []
countries = []
values = []
stat_names = []

for i in range(len(team_data)):
    top_three = team_data[i]['topThree']
    for j in range(len(top_three)):
        names.append(top_three[j]['name'])
        ranks.append(top_three[j]['rank'])
        countries.append(top_three[j]['ccode'])
        values.append(top_three[j]['value'])
        stat_names.append(team_data[i]['header'])

team_df = pd.DataFrame({'Team': names,
             'League Rank': ranks,
             'Country': countries,
             'Value': values,
             'Stat Name': stat_names})
```

```python
xg_conceded_url = 'https://www.fotmob.com/api/data/leagueseasondeepstats?lng=en-GB&id=48&season=27195&type=teams&stat=expected_goals_conceded_team'
xg_faced_response = requests.get(xg_conceded_url, headers=headers)
```

```python
goals_conceded_url = 'https://www.fotmob.com/api/data/leagueseasondeepstats?lng=en-GB&id=48&season=27195&type=teams&stat=goals_conceded_team_match'
goals_conceded_response = requests.get(goals_conceded_url, headers=headers)
```

```python
goals_conceded_data = goals_conceded_json['statsData']

goals_conceded_data[0]
```

```python
teams = []
values = []
for i in range(len(goals_conceded_data)):
    teams.append(goals_conceded_data[i]['name'])
    values.append(goals_conceded_data[i]['substatValue']['value'])

goals_conceded_df = pd.DataFrame({'team': teams,
                                  'goals_conceded': values})

goals_conceded_df
```

```python
teams = []
values = []
for i in range (len(xg_faced_data)):
    teams.append(xg_faced_data[i]['name'])
    values.append(xg_faced_data[i]['statValue']['value'])

xg_faced_df = pd.DataFrame({'team': teams,
                            'xg_faced': values})
```

```python
defence_quality_df = pd.merge(goals_conceded_df, xg_faced_df, how='left', on='team')

defence_quality_df
```
